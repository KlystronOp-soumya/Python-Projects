##### Function in python are first class objects
- It is possible to assign a function to a variable
- It is also possible to define one function inside another function
- Can be passed as a parameter to another function
- function can return another function

```python
def display(name) :
    
    # function can be defined inside a function as well    
    def message():
        return "Howdy!"
    
    def getName():
        return "Python"
    # can also return a function
    #return getName  
    return message() + name
    #return "Hi" + name

def foo():
    return "This is Foo"

def bar(foo) :
    return "Greetings !!" + foo

#assigning to a variable
x = display("abd")
print(x)
print(bar(foo()))
```

##### Pass By Value and Reference

##### types of function arguments
- Positional Arguments
- Keyword arguments
- Default Arguments
- Varibale length Arguments

```python
#positional Arguments
def showAddress(address , zip , area , state):
    print(" address: {} \n zip: {} \n area: {} \n state: {}".format(address , zip , area , state))

def simpleInterest(p , r=6.7, t= 2) :
    print(p*r*t / 100)

#variable length arguments
def sumOfNumbers(farg , *args):
    """ 
    farg : formal argument
    *args: variable length argument 
    """
    
    print("Passed formal argument is : " , farg)
    
    sum = 0
    
    for i in args :
        sum+=i
    print(type(args))
    print(args)
    print("Sum of numbers : " , sum)
    
    return

#variable length keyword argumnets
def showStudentDetails(farg , **kwargs):
    """_summary_
       farg: formal arguments
       **kwargs: keyword variable length argument
    Args:
        farg (_type_): _description_
    """
    print("Passed formal arguments: " , farg)
    print(kwargs)
    for key , val in kwargs.items():
        print(key , ":" , val)
     
showAddress("colony more" , "700126" , "Barasat" , "WB")
showAddress("colony more" , "700126" , "WB" , "barasat")
simpleInterest(10000 , 5 , 3)
#Problem with the positional arguments is to maintain the integrity and consistency

simpleInterest(p=15000 , r = 10 , t = 5)
simpleInterest(t=10 , p = 500000 , r = 19.6)
simpleInterest(p=10000 , r = 10 , t= 5)
simpleInterest(1000) #valid
simpleInterest(p=1000) #valid
#simpleInterest() #invalid -- it requires at least one parameter

sumOfNumbers(4 , 1 ,2,3 ,4 , 5 ,6,7)
sumOfNumbers(4 , 1 ,2,3)

showStudentDetails(3 , studentName="Myra" , studentClass = "12" , studentSec="B")
showStudentDetails(3 , studentName="Myra" , studentClass = "12" , studentSec="B" , rollNo="13" , boardsId="BST1234")
```
