##### Simple program to show python File Handling

```python
#Conventional way of opening and closing a file in python
f = open(
    file= "lorem.txt" ,
    mode="+a" ,
    buffering= 1024
)

f.close() #if we are opening any stream , we have to close it
```

```python
with open(file="Documents\lorem.txt" , mode="+r" , encoding= "UTF-8") as contents:
   lines = contents.readlines()
   print(lines)
 # here we do not need to close the stream explicitly   
```

Python program to show the contents in a file

```python
#now show the contents of the file we created:
print("\n Showing the contents of the file")
with open("Documents\\lorem.txt" , "r" , 1024)  as myFile :
    fileContents = myFile.read()
    print(fileContents)

```

```python
#Another way to display the contents of the files
#now show the contents of the file we created:
print("\n Showing the contents of the file")
with open("Documents\\lorem.txt" , "r" , 1024)  as myFile :
    for eachLine in myFile.readlines:
        print(eachLine)
        
```

Check if a file exists or not

```python
#first import two built-in modules in python
import os , sys

# get the file name from the user
fileName = "lorem.txt"

if os.path.isfile(fileName):
    file = open(fileName , 'r+')
else:
    print("File: " , fileName , " does not exist")
    sys.exit()

print("The file contents are")
fileContents = file.read()
print(fileContents )
```

##### Binary files in Python

```python
import sys
str="Python"

try:
    with open("data.bin" , "wb+" ) as myBinFile:
        myBinFile.write(str.encode("utf-8"))
        
    # now read the entry
    with open("data.bin" , "rb") as myBinFile:
        str = myBinFile.read()
        print(str.decode())
except Exception as e :
    print("Message: " , e , file= sys.stderr)
   
```

#### Inserting data into a binary file without pickling

```python
RECLEN = 20
cityList = ["Pune" , "Hyderabad" , "Goa" , "Kolkata" , "Mumbai"]
with open("cities.bin" , "wb") as f :
    #cityName = input("Enter city name")
    for cityName in cityList :
        #get the length of the city
        cityNmLn = len(cityName)
        #pad the name with spaces to maintain the length as 20 chars
        print(f"#Debug : name scanned {cityName} length {cityNmLn} ")
        cityName = cityName + (RECLEN - cityNmLn)*' '
        #convert the string into binary
        cityEncoded = cityName.encode()
        #write into the file
        f.write(cityEncoded)
```

```python
#Reading random entries from Binary file
RECLEN = 20

with open("cities.bin" , "rb") as f :
    #cityName = input("Enter city name")
    recordNumber = 2
    # move the cursor or the file pointer to the end of n-1 record
    f.seek(RECLEN * (recordNumber - 1))
    # read the ninary data
    cityName = f.read(RECLEN)
    #decode the binary data into character data
    print(cityName.decode())
```

##### Seek and Tell in Python: Controlling the cursor

```python
import os

with open("line.txt" , "w+b") as f :
    f.write("Great Python".encode("utf-16"))

with open ("line.txt" , "r+b") as f :
    print("current cursor position: " , f.tell())
    f.seek(5)
    print("current cursor position: " , f.tell())
    print(f.read(5).decode())
    print("current cursor position: " , f.tell())
    print(f.read(5).decode())
    print("current cursor position: " , f.tell())
    f.seek(-5 , 2)
    print("current cursor position: " , f.tell())
    print(f.read(5).decode())
    print("current cursor position: " , f.tell())
```

```python
import struct
str="Python"
version = 3.7
float_to_bytes = lambda value: struct.pack('d', value)
float_bytes = float_to_bytes(version)
print(float_bytes)
try:
    with open("data.dat" , "wb+" ) as myBinFile:
        myBinFile.write(str.encode())
        myBinFile.write("\n".encode())
        myBinFile.write(bytes(float_bytes))
except Exception as e :
    print(e , sys.stdout)
```
